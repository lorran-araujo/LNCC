```python
import os
from PIL import Image
from tqdm import tqdm

import torch
from torchvision.datasets import OxfordIIITPet
from torchvision import transforms
from torch.utils.data import DataLoader
from torch.utils.data import Dataset
```

```python
# =========================
# Configurações
# =========================

IMG_SIZE = 128

RAW_DIR = "./data/raw"
PROCESSED_DIR = "./data/oxford_pets_128"

os.makedirs(PROCESSED_DIR, exist_ok=True)


# =========================
# Dataset original
# =========================

dataset = OxfordIIITPet(
    root=RAW_DIR,
    split="trainval",
    target_types="category",
    download=True
)


print(f"Número de imagens: {len(dataset)}")
```

```python
# =========================
# Transformação
# =========================

transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.CenterCrop(IMG_SIZE),
    transforms.ToTensor(),

    # Difusão normalmente trabalha em [-1,1]
    transforms.Normalize(
        mean=[0.5, 0.5, 0.5],
        std=[0.5, 0.5, 0.5]
    )
])


# =========================
# Processamento
# =========================

for idx, (img, label) in enumerate(tqdm(dataset)):

    img = transform(img)

    # volta para imagem para salvar
    img_save = img * 0.5 + 0.5
    img_save = torch.clamp(img_save, 0, 1)

    img_save = transforms.ToPILImage()(img_save)

    filename = f"{idx:05d}.png"

    img_save.save(
        os.path.join(PROCESSED_DIR, filename)
    )


print("Pré-processamento finalizado!")
print(f"Imagens salvas em: {PROCESSED_DIR}")
```

```python
from torchvision.datasets import CIFAR100
import os
from tqdm import tqdm

output_dir = "cifar100_images"

dataset = CIFAR100(
    root="./data",
    train=True,
    download=True
)

os.makedirs(output_dir, exist_ok=True)

for i in tqdm(range(len(dataset))):
    image, _ = dataset[i]
    image.save(os.path.join(output_dir, f"{i:05d}.png"))

print("Imagens salvas!")
```

```python
from torchvision.datasets import MNIST
import os
from tqdm import tqdm

# Pasta onde as imagens serão salvas
output_dir = "mnist_images"

# Baixa o dataset
dataset = MNIST(
    root="./data",
    train=True,
    download=True
)

os.makedirs(output_dir, exist_ok=True)

for i in tqdm(range(len(dataset)), desc="Salvando imagens"):

    image, _ = dataset[i]  # ignora o rótulo

    image.save(
        os.path.join(
            output_dir,
            f"{i:05d}.png"
        )
    )

print(f"{len(dataset)} imagens salvas em '{output_dir}'!")
```
