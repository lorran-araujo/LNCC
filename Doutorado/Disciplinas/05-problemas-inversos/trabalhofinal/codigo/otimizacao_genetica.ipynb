# Calibracao por algoritmo genetico em lotes

As combinacoes ja concluidas sao ignoradas. Cada bloco grava seu proprio CSV em `dados/resultados/lotes/` e atualiza `metricas.csv` sem duplicar resultados.

```python
from pathlib import Path

from genetic_batch import BatchConfig, run_batch
from genetic_calibration import GeneticAlgorithmConfig

data_dir = Path("..") / "dados"

# Execute blocos consecutivos: 0-10, 10-20, ..., 130-140.
case_start = 0
case_stop = 10  # Exclusivo. Este bloco executa os case_id de 0 ate 9.
workers = 4

batch_config = BatchConfig(
    case_ids=tuple(range(case_start, case_stop)),
    noise_fractions=(0.01, 0.05, 0.10),
    noise_replicates=(0, 1, 2, 3, 4),
    optimizer_seeds=(2027, 2028, 2029, 2030, 2031),
    workers=workers,
)
algorithm_config = GeneticAlgorithmConfig(population_size=80, generations=150)

expected_runs = (
    len(batch_config.case_ids)
    * len(batch_config.noise_fractions)
    * len(batch_config.noise_replicates)
    * len(batch_config.optimizer_seeds)
)
print(f"Combinacoes solicitadas neste bloco: {expected_runs}")
```

```python
summary = run_batch(
    data_dir / "piloto_trajectories.npz",
    data_dir / "resultados",
    batch_config,
    algorithm_config,
)

print(f"Solicitadas: {summary.requested}")
print(f"Concluidas agora: {summary.completed}")
print(f"Ignoradas por ja existirem: {summary.skipped}")
print(f"Lote: {summary.batch_path}")
print(f"Metricas consolidadas: {summary.metrics_path}")
```
